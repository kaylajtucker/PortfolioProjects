# **Slide 1: Problem Statement**

For this project, we decided to analyze the Adult Census Income dataset. The goal is to explore demographic and work-related factors to better understand the characteristics that may influence whether an individual earns more than $50K per year in the 1990s. The dataset comes from the 1994 U.S. Census and contains over 30,000 records with 15 attributes, including age, education, occupation, work hours, and income level. This dataset is widely used for classification problems, but we focusing on exploratory data analysis to uncover patterns, distributions, and relationships in the data.

# **Slide 4**

```python
#Univariate Analysis – Categorical variables.
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="income")
plt.title("Income Distribution (Target Variable)")
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, y="workclass", order=df['workclass'].value_counts().index)
plt.title("Distribution of Workclass")
plt.show()
```

```python
#Univariate Analysis – Continuous Variables.
plt.figure(figsize=(8,5))
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['hours.per.week'], bins=30, kde=True)
plt.title("Hours per Week Distribution")
plt.show()
```

```python
#Bivariate Analysis - Scatter Plot.
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="age", y="hours.per.week", hue="income", alpha=0.5)
plt.title("Age vs Hours per Week (colored by Income)")
plt.show()
```

```python
#Data Skew / Class imbalance.
print("Income class distribution:")
print(df['income'].value_counts(normalize=True))

sns.countplot(x='income', data=df)
plt.title('Income Class Distribution')
plt.show()
```

```python
#Linearity vs non-linearity between features.
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="education.num", y="hours.per.week", hue="income", alpha=0.5)
plt.title("Education Level vs Hours per Week")
plt.show()
```

```python
#the plot above aids my observation: people who are getting paid are working over 40 hours, way more than people making over 50k
print(len(df.loc[(df['income'] == '>50K') & (df['hours.per.week']>40)]))

len(df.loc[(df['income'] == '<=50K') & (df['hours.per.week']>40)])
```

```python
#Box plot: age
plt.figure(figsize=(8,5))
sns.boxplot(x=df['age'])
plt.title("Boxplot of Age")
plt.show()
```

```python
#Box plot: hours per week
plt.figure(figsize=(8,5))
sns.boxplot(x=df['hours.per.week'])
plt.title("Boxplot of Hours per Week")
plt.show()
```

```python
#Violin plot: age by income
plt.figure(figsize=(8,5))
sns.violinplot(data=df, x="income", y="age")
plt.title("Age by Income (Violin Plot)")
plt.show()
```

# **Slide 6**

```python
plt.figure(figsize=(10,6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Numeric Features")
plt.show()
```

# **Extra**

```python
#income by education
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="education", hue="income", order=df['education'].value_counts().index)
plt.xticks(rotation=45)
plt.title("Income Distribution by Education Level")
plt.show()
```

```python
#income by occupation
plt.figure(figsize=(12,6))
sns.countplot(data=df, y="occupation", hue="income", order=df['occupation'].value_counts().index)
plt.title("Income Distribution by Occupation")
plt.show()
```

```python
#income by marital status
plt.figure(figsize=(10,5))
sns.countplot(data=df, x="marital.status", hue="income", order=df['marital.status'].value_counts().index)
plt.xticks(rotation=45)
plt.title("Income Distribution by Marital Status")
plt.show()
```

```python
#race and sex vs income
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="race", hue="income", order=df['race'].value_counts().index)
plt.title("Income Distribution by Race")
plt.show()

plt.figure(figsize=(6,5))
sns.countplot(data=df, x="sex", hue="income")
plt.title("Income Distribution by Sex")
plt.show()
```

```python
cat_pivot = pd.crosstab(df['education'], df['income'], normalize='index')
plt.figure(figsize=(8,6))
sns.heatmap(cat_pivot, annot=True, cmap="Blues", fmt=".2f")
plt.title("Proportion of Income Levels by Education")
plt.show()
```
